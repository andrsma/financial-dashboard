In 2026, a region has 50 billion m3 of available freshwater. Each year, natural replenishment adds 2 billion m3, while agriculture and other sectors withdraw 2.5 billion m3

```python
import numpy as np
import matplotlib.pyplot as plt

#In 2026 a region has 50 billion m3 of freshwater.
# Each year 2 billion are added naturally
#Each year agriculture and other sectors withdraw 2.5 billion

#Formula for water available after x years, with year 0 being 2026: y = -0.5x + 50 (in billions m3)
years = np.arange(2026,2101) #last year is not inclusive
#1
stock = 50 - 0.5 * np.arange(len(years))

#2
below_30 = years[stock < 30][0] #From all years where stock is below 30, select the first one
print(below_30)

plt.plot(years, stock)
plt.title('Freshwater Stock 2026-2100')
plt.xlabel("Years")
plt.ylabel(r"Volume ($m^3$) in billions")
plt.axhline(30, color="gray", linestyle="--", label=r"30 $m^3$")
plt.axvline(below_30, color="red", linestyle=":")
plt.annotate(f"First year below 30: {below_30}", xy=(below_30, np.where(years==below_30)[0][0]),
                 xytext=(10, 20), textcoords="offset points",
                 arrowprops=dict(arrowstyle="->", color="red"))
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
